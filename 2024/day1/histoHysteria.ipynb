```python
with open('input.txt') as file:
    locationIDs = file.read().strip(" ").split("\n")
```

```python
left = []
right = []
for i in locationIDs:
    left.append(int(i.split("   ")[0]))
    right.append(int(i.split("   ")[1]))

similarityScore = 0
for i in left:
    occuranceCount = right.count(i)
    similarityScore += occuranceCount * i
print('similarityScore is', similarityScore)

left.sort()
right.sort()

sum = 0
for i in range(len(left)):
    sum += abs(left.pop(0) - right.pop(0))

print('totalDistance is', sum)
```
