```python
class Node:
    def __init__(self, value: int):
        self.value = value
        self.next: "Node | None" = None

def build_circular(start: int = 50) -> Node:
    head = Node(start)
    prev = head
    for i in range(1, 100):
        val = (start + i) % 100
        node = Node(val)
        prev.next = node
        prev = node
    prev.next = head  # close the circle
    return head

def move(head: Node, steps: int) -> Node:
    # steps can be positive (forward) or negative (simulate backward)
    size = 100  # it’s 0..99
    steps = steps % size
    cur = head
    for _ in range(steps):
        cur = cur.next
    return cur

head = build_circular(50)  # head.value == 50
```

```python
# number of times it stops at 0
head = build_circular(50)
counter = 0
for i in rotations:
    direction = i[0]
    rotation = int(i[1:])
    if direction == "L":
        head = move(head, -rotation)
        if head.value == 0:
            counter += 1
    if direction == "R":
        head = move(head, rotation)
        if head.value == 0:
            counter += 1
print(counter)
        
```

```python
from pathlib import Path

def count_zero_crossings(path, start=50, size=100):
    pos = start % size
    hits = 0
    for raw in Path(path).read_text().split():
        if not raw:
            continue
        direction, steps = raw[0], int(raw[1:])
        if direction == "R":
            first = (size - pos) % size or size  
            if steps >= first:
                hits += 1 + (steps - first) // size
            pos = (pos + steps) % size
        else:
            first = pos or size
            if steps >= first:
                hits += 1 + (steps - first) // size
            pos = (pos - steps) % size
    return hits, pos

if __name__ == "__main__":
    crossings, final_pos = count_zero_crossings("rotations.txt")
    print(f"Zero crossings: {crossings}")
    print(f"Final position: {final_pos}")
```
