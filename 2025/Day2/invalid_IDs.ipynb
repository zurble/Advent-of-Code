```python
from pathlib import Path
```

```python
def count_zero_crossings(path):
    for raw in Path(path).read_text().split(","):
        first,second = raw.split("-")
        if len(first)%2!=0 and len(second)%2!=0:
            continue
        

if __name__ == "__main__":
    solution = count_zero_crossings("input.txt")
```
